```python
# ============================================================
# STEP 1: Import Required Libraries
# Project: Cricsheet Match Data Analysis
# ============================================================

import os
import json
import sqlite3
import pandas as pd

print(" All required libraries imported successfully.")
```

```python
# ============================================================
# STEP 2: Validate Project Folder Structure
# ============================================================

import os

project_path = os.getcwd()
data_path = os.path.join(project_path, "data")

folders = [
    "ipl_json",
    "odis_json",
    "tests_json",
    "it20s_json"
]

for folder in folders:
    folder_path = os.path.join(data_path, folder)

    if os.path.exists(folder_path):
        total_files = len([file for file in os.listdir(folder_path) if file.endswith(".json")])
        print(f"{folder} : {total_files} JSON files")
    else:
        print(f"{folder} : Folder not found")
```

```python
# ============================================================
# STEP 3: Create SQLite Database Connection
# ============================================================

import sqlite3

database_name = "cricsheet.db"

conn = sqlite3.connect(database_name)
cursor = conn.cursor()

print("Database connected successfully.")
```

```python
# ============================================================
# STEP 4: Verify Database Connection
# ============================================================

cursor.execute("SELECT sqlite_version();")

version = cursor.fetchone()

print("SQLite Version:", version[0])
```
