```python
# Selenium Libraries Import

import os
import time

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
```

```python
# Download Folder Create

download_folder = os.path.join(os.getcwd(), "downloads")
os.makedirs(download_folder, exist_ok=True)

print(download_folder)
```

```python
options = Options()

prefs = {
    "download.default_directory": download_folder,
    "download.prompt_for_download": False,
    "download.directory_upgrade": True
}

options.add_experimental_option("prefs", prefs)

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)

driver.maximize_window()
driver.get("https://cricsheet.org/matches/")

print("Website Opened Successfully ✅")
```

```python
links = driver.find_elements(By.TAG_NAME, "a")

for link in links:
    text = link.text.strip()
    if text:
        print(text)
```

```python
tables = driver.find_elements(By.TAG_NAME, "table")

print("Total Tables:", len(tables))
```

```python
import zipfile
import os

extract_folder = os.path.join(os.getcwd(), "data")
os.makedirs(extract_folder, exist_ok=True)

zip_files = [
    "ipl_json.zip",
    "tests_json.zip",
    "odis_json.zip",
    "it20s_json.zip"
]

for zip_file in zip_files:
    zip_path = os.path.join(download_folder, zip_file)

    if os.path.exists(zip_path):
        folder_name = zip_file.replace(".zip", "")
        destination = os.path.join(extract_folder, folder_name)
        os.makedirs(destination, exist_ok=True)

        with zipfile.ZipFile(zip_path, "r") as z:
            z.extractall(destination)

        print(f"{zip_file} Extracted ✅")
    else:
        print(f"{zip_file} Not Found ❌")
```
