```python
# ============================================================
# STEP 1: Import Libraries & Connect SQLite Database
# ============================================================

import sqlite3
import pandas as pd

# Connect to SQLite Database
conn = sqlite3.connect("cricsheet.db")
cursor = conn.cursor()

print("Database Connected Successfully!")
```

```python
# ============================================================
# STEP 2: Verify Connected Database
# ============================================================

cursor.execute("PRAGMA database_list;")
print(cursor.fetchall())
```

```python
# ============================================================
# STEP 3: Check Available Tables
# ============================================================

cursor.execute("""
SELECT name
FROM sqlite_master
WHERE type='table';
""")

print(cursor.fetchall())
```

```python
# ============================================================
# STEP 4: SQL Query 1 - Total Number of Matches
# ============================================================

query = """
SELECT COUNT(*) AS total_matches
FROM matches;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 5: SQL Query 2 - Matches by Match Type
# ============================================================

query = """
SELECT
    match_type,
    COUNT(*) AS total_matches
FROM matches
GROUP BY match_type
ORDER BY total_matches DESC;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 6: SQL Query 3 - Top 10 Teams by Matches Played
# ============================================================

query = """
SELECT
    team,
    COUNT(*) AS matches_played
FROM
(
    SELECT team1 AS team FROM matches
    UNION ALL
    SELECT team2 AS team FROM matches
)
GROUP BY team
ORDER BY matches_played DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 7: SQL Query 4 - Top 10 Teams by Wins
# ============================================================

query = """
SELECT
    winner,
    COUNT(*) AS total_wins
FROM matches
WHERE winner IS NOT NULL
  AND winner <> ''
GROUP BY winner
ORDER BY total_wins DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 8: SQL Query 5 - Top 10 Venues by Matches Hosted
# ============================================================

query = """
SELECT
    venue,
    COUNT(*) AS total_matches
FROM matches
GROUP BY venue
ORDER BY total_matches DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 9: SQL Query 6 - Top 10 Player of the Match Winners
# ============================================================

query = """
SELECT
    player_of_match,
    COUNT(*) AS awards
FROM matches
WHERE player_of_match IS NOT NULL
  AND player_of_match <> ''
GROUP BY player_of_match
ORDER BY awards DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 10A: Check Deliveries Table Columns
# ============================================================

cursor.execute("PRAGMA table_info(deliveries);")

columns = cursor.fetchall()

for column in columns:
    print(column)
```

```python
# ============================================================
# STEP 10: SQL Query 7 - Top 10 Run Scorers
# ============================================================

query = """
SELECT
    batter,
    SUM(runs_batter) AS total_runs
FROM deliveries
GROUP BY batter
ORDER BY total_runs DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 11: SQL Query 8 - Top 10 Wicket Takers
# ============================================================

query = """
SELECT
    bowler,
    COUNT(*) AS wickets
FROM deliveries
WHERE wicket = 1
GROUP BY bowler
ORDER BY wickets DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 12: SQL Query 9 - Top 10 Teams by Total Runs
# ============================================================

query = """
SELECT
    batting_team,
    SUM(runs_total) AS total_runs
FROM deliveries
GROUP BY batting_team
ORDER BY total_runs DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 13: SQL Query 10 - Highest Team Innings Scores
# ============================================================

query = """
SELECT
    match_id,
    batting_team,
    innings,
    SUM(runs_total) AS total_score
FROM deliveries
GROUP BY match_id, batting_team, innings
ORDER BY total_score DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 14: SQL Query 11 - Toss Decision Analysis
# ============================================================

query = """
SELECT
    toss_decision,
    COUNT(*) AS total_matches
FROM matches
GROUP BY toss_decision
ORDER BY total_matches DESC;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 15: SQL Query 12 - Matches Played Per Year
# ============================================================

query = """
SELECT
    SUBSTR(match_date, 1, 4) AS year,
    COUNT(*) AS total_matches
FROM matches
GROUP BY year
ORDER BY year;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 16: SQL Query 13 - Average Runs Per Match
# ============================================================

query = """
SELECT
    ROUND(AVG(match_runs), 2) AS average_runs_per_match
FROM (
    SELECT
        match_id,
        SUM(runs_total) AS match_runs
    FROM deliveries
    GROUP BY match_id
);
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 17: SQL Query 14 - Toss Winner vs Match Winner
# ============================================================

query = """
SELECT
    CASE
        WHEN toss_winner = winner THEN 'Won Match'
        ELSE 'Lost Match'
    END AS toss_result,
    COUNT(*) AS matches
FROM matches
WHERE winner IS NOT NULL
  AND winner <> ''
GROUP BY toss_result;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 18A: Check Matches Table Columns
# ============================================================

cursor.execute("PRAGMA table_info(matches);")

columns = cursor.fetchall()

for column in columns:
    print(column)
```

```python
# ============================================================
# STEP 18: SQL Query 15 - Win Type Analysis
# ============================================================

query = """
SELECT
    CASE
        WHEN win_by_runs > 0 THEN 'Won by Runs'
        WHEN win_by_wickets > 0 THEN 'Won by Wickets'
        ELSE 'Other / Tie / No Result'
    END AS win_type,
    COUNT(*) AS total_matches
FROM matches
GROUP BY win_type
ORDER BY total_matches DESC;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 19: SQL Query 16 - Matches by Gender
# ============================================================

query = """
SELECT
    gender,
    COUNT(*) AS total_matches
FROM matches
GROUP BY gender
ORDER BY total_matches DESC;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 20: SQL Query 17 - Top 10 Cities by Matches Hosted
# ============================================================

query = """
SELECT
    city,
    COUNT(*) AS total_matches
FROM matches
WHERE city IS NOT NULL
  AND city <> ''
GROUP BY city
ORDER BY total_matches DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 21: SQL Query 18 - Teams Winning by 100+ Runs
# ============================================================

query = """
SELECT
    winner,
    COUNT(*) AS matches_won
FROM matches
WHERE win_by_runs >= 100
  AND winner IS NOT NULL
  AND winner <> ''
GROUP BY winner
ORDER BY matches_won DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 22: SQL Query 19 - Teams Winning by 10 Wickets
# ============================================================

query = """
SELECT
    winner,
    COUNT(*) AS matches_won
FROM matches
WHERE win_by_wickets = 10
  AND winner IS NOT NULL
  AND winner <> ''
GROUP BY winner
ORDER BY matches_won DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 23: SQL Query 20 - Match Type vs Gender
# ============================================================

query = """
SELECT
    match_type,
    gender,
    COUNT(*) AS total_matches
FROM matches
GROUP BY match_type, gender
ORDER BY match_type, gender;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ============================================================
# STEP 24: Close Database Connection
# ============================================================

conn.close()

print("Database connection closed successfully.")
```
