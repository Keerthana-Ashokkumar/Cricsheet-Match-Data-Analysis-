```python
# Step 1: Import Required Libraries

import sqlite3
import pandas as pd
import os

print("Libraries imported successfully.")
```

```python
# Step 2: Connect to the SQLite Database

db_path = "cricsheet.db"

conn = sqlite3.connect(db_path)

print("Connected to SQLite database successfully.")
```

```python
# Step 3: Load Required Tables from SQLite

matches_df = pd.read_sql_query("SELECT * FROM matches", conn)
deliveries_df = pd.read_sql_query("SELECT * FROM deliveries", conn)

print("Tables loaded successfully.")

print("\nMatches Shape:", matches_df.shape)
print("Deliveries Shape:", deliveries_df.shape)
```

```python
# Step 4: Create Output Folder for Power BI Files

output_folder = "csv"

os.makedirs(output_folder, exist_ok=True)

print(f"Output folder '{output_folder}' is ready.")
```

```python
# Step 5: Export the Matches Table to CSV

matches_file = os.path.join(output_folder, "matches.csv")

matches_df.to_csv(matches_file, index=False)

print(f"Matches table exported successfully to: {matches_file}")
```

```python
# Step 6: Export the Deliveries Table to CSV

deliveries_file = os.path.join(output_folder, "deliveries.csv")

deliveries_df.to_csv(deliveries_file, index=False)

print(f"Deliveries table exported successfully to: {deliveries_file}")
```

```python
# Step 7: Verify the Exported CSV Files

print("Files available in the 'csv' folder:")

for file in os.listdir(output_folder):
    print(file)
```

```python
# Step 8: Close the Database Connection

conn.close()

print("Database connection closed successfully.")
```
